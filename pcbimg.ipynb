```python
import openai 
import json
```

```python
pcb_data = {
    "net": "GND",
    "trace_width_mil": 4,
    "min_trace_width_mil": 6,
    "clearance_mil": 5,
    "min_clearance_mil": 6
}

img_path = "pcbimage.png"  # Upload your PCB image here
```

```python
def drc_check(pcb):
    errors = []
    if pcb["trace_width_mil"] < pcb["min_trace_width_mil"]:
        errors.append(f"Trace width {pcb['trace_width_mil']} mil < minimum {pcb['min_trace_width_mil']} mil")
    if pcb["clearance_mil"] < pcb["min_clearance_mil"]:
        errors.append(f"Clearance {pcb['clearance_mil']} mil < minimum {pcb['min_clearance_mil']} mil")
    return errors

print("DRC Errors:", drc_check(pcb_data))
```

```python
import json


prompt = f"""
You are a PCB design checker.
Given the structured data: {json.dumps(pcb_data)},
and the uploaded PCB image (visual context),
1. Identify rule violations.
2. Suggest corrective actions (e.g., increase trace width, adjust clearance).
3. Explain why the correction is necessary.
"""

response = openai.chat.completions.create(
    model="gpt-4.1",  # multimodal model
    messages=[
        {"role":"user","content":prompt},
        {"role":"user","content":[
            {"type":"text","text":"Here is the PCB image for context:"},
            {"type":"image_url","image_url":{"url":"pcb_example.png"}}
        ]}
    ]
)

print("LLM Output:", response.choices[0].message.content)
```
