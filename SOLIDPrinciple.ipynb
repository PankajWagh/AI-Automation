# -------------------------------
# 1. Single Responsibility Principle (SRP)
# -------------------------------
# Each class has one responsibility:
# - Order handles order details
# - PaymentProcessor handles payment logic

```python
from abc import ABC, abstractmethod
class Order:
    def __init__(self, order_id: str, amount: float):
        self.order_id = order_id
        self.amount = amount
```

# -------------------------------
# 2. Open/Closed Principle (OCP)
# -------------------------------
# PaymentProcessor is open for extension (new payment methods)
# but closed for modification (no need to change existing code).

```python
class PaymentProcessor(ABC):
    @abstractmethod
    def pay(self, order: Order) -> None:
        pass


class CreditCardProcessor(PaymentProcessor):
    def pay(self, order: Order) -> None:
        print(f"Processing credit card payment for order {order.order_id}, amount {order.amount}")


class PayPalProcessor(PaymentProcessor):
    def pay(self, order: Order) -> None:
        print(f"Processing PayPal payment for order {order.order_id}, amount {order.amount}")
```

# -------------------------------
# 3. Liskov Substitution Principle (LSP)
# -------------------------------
# Any subclass of PaymentProcessor can replace the parent without breaking behavior.

```python
def process_order(order: Order, processor: PaymentProcessor):
    processor.pay(order)
```

# -------------------------------
# 4. Interface Segregation Principle (ISP)
# -------------------------------
# Instead of one giant interface, we define smaller, role-specific ones.

```python
class Refundable(ABC):
    @abstractmethod
    def refund(self, order: Order) -> None:
        pass


class RefundableCreditCardProcessor(CreditCardProcessor, Refundable):
    def refund(self, order: Order) -> None:
        print(f"Refunding credit card payment for order {order.order_id}, amount {order.amount}")
```

# -------------------------------
# 5. Dependency Inversion Principle (DIP)
# -------------------------------
# High-level modules depend on abstractions, not concrete implementations.

```python
class CheckoutService:
    def __init__(self, processor: PaymentProcessor):
        self.processor = processor

    def checkout(self, order: Order):
        self.processor.pay(order)

```

# -------------------------------
# Example Usage
# -------------------------------

```python
order = Order("123", 250.0)
# Inject dependency (DIP)
checkout = CheckoutService(CreditCardProcessor())
checkout.checkout(order)

checkout = CheckoutService(PayPalProcessor())
checkout.checkout(order)
```
