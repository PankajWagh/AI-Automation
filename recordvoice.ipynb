```python
import sounddevice as sd
import numpy as np
import wave

def record_audio(filename="input.wav", duration=5, samplerate=16000):
    print("Recording...")
    audio = sd.rec(int(duration * samplerate), samplerate=samplerate, channels=1, dtype='int16')
    sd.wait()
    wave_file = wave.open(filename, 'wb')
    wave_file.setnchannels(1)
    wave_file.setsampwidth(2)
    wave_file.setframerate(samplerate)
    wave_file.writeframes(audio.tobytes())
    wave_file.close()
    print("Saved:", filename)

record_audio("input.wav", duration=5)
```

```python
from openai import OpenAI
client = OpenAI()

with open("input.wav", "rb") as f:
    transcription = client.audio.transcriptions.create(
        model="gpt-4o-mini-transcribe", 
        file=f
    )

print("Transcribed text:", transcription.text)
```

```python
translation = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a translator that converts English text into Marathi."},
        {"role": "user", "content": transcription.text}
    ]
)

print("Marathi translation:", translation.choices[0].message.content)
```
