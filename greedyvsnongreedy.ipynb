The core difference is that greedy quantifiers match as much text as possible, 
while non-greedy (lazy) quantifiers match as little text as possible to make the overall pattern succeed.
In Python's built-in re module, standard repetitions (like *, +, and ?) default to greedy behavior.
Adding a question mark (?) right after a quantifier instantly switches it to a non-greedy mode.

```python
import re 

TEXT  = "<div>Hello</div> <div>World</div>"

# Greedy: .* will consume as much as possible
greedy_match = re.findall(r"<div>.*</div>", TEXT )
print(greedy_match)
# Output: ['<div>Hello</div> <div>World</div>']

# Non-Greedy (Lazy): .*? will stop at the first valid closing </div>
lazy_match = re.findall(r"<div>.*?</div>", TEXT )
print(lazy_match)
# Output: ['<div>Hello</div>', '<div>World</div>']
```
