ReAct (Reason + Act) is a prompting pattern where the model alternates between:
Reasoning — thinking step by step about the problem.
Acting — calling tools, APIs, or functions to gather information.
Observing — reading the tool’s output.
Continuing — reasoning again until the final answer.
This is especially useful in LLM agents that need both reasoning and external actions.

```python
import openai

response = openai.ChatCompletion.create(
    model="gpt-4",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Use ReAct reasoning to answer: What is 12 * 7?"}
    ]
)

print(response.choices[0].message["content"])
```

```python
import openai

def calculator(expression):
    return eval(expression)

prompt = """
Question: What is 12 * 7?
Use ReAct: reason, act (call calculator), observe, answer.
"""

response = openai.ChatCompletion.create(
    model="gpt-4.1",
    input=[
        {"role": "system", "content": "You can call tools like calculator."},
        {"role": "user", "content": prompt}
    ],
    tools=[{"name": "calculator"}],  # tool registration
    max_tokens=150,
)

print(response.choices[0].message["content"])
```
